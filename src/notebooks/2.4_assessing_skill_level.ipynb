```python
from IPython.display import display
import sys
sys.path.insert(0, "../fretwave")
import librosa
import chord_detect
```

```python
clip, sr = librosa.load("../data/clips/sample/all_chords.wav", sr=None)
```

```python
detected = chord_detect.get_chords(clip, sr)
```

```python
# a dictionary of dictonaries contianing skill levels and the modifiers for each skill level
skill_levels = {
    "beginner" : {
        "chord_accuracy_threshold": 60,
        "transition_target": None,
    },
    "intermediate" : {
        "chord_accuracy_threshold": 75,
        "transition_target": None,
    },
    "advanced": {
        "chord_accuracy_threshold": 85,
        "transition_target": None,
    },
}
```

```python
def chord_accepted(level, chord_accuracy):
    if chord_accuracy < skill_levels[level]["chord_accuracy_threshold"]:
        print(f"Chord accuracy {chord_accuracy} is below the threshold for {level} guitar players ({skill_levels[level]["chord_accuracy_threshold"]})")
    else:
        print("Chord detected!")
```

```python
level = "advanced"
```

Taking the `Chord: D, Score:75.52%` entry from the detected chords, (3rd entry, index = 2) as this tests if the function will return false for an advanced player, but true for an intermediate (whilst also acting on the near boundary for intermediate, where the threshold is 75)

```python
chord_score = detected[2][1]
```

```python
chord_accepted(level, chord_score)
```

```python
level = "intermediate"
chord_accepted(level, chord_score)
```
