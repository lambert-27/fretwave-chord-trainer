This notebook aims to prompt the user with three images of chords represented as tab diagrams. The goal is to prompt the user with three chords, have them record the three chord changes and perform analysis



```python
# Source - https://stackoverflow.com/a/26392492
# Posted by user1269942
# Retrieved 2026-10-21, License - CC BY-SA 3.0

from PIL import Image
from pathlib import Path
import glob
# chord dict, so name can be tracked and matched with template
chords = {}
for filename in glob.glob('chords/*.jpg'):
    # need pathlib to get the filename, for analysis use -> matching against tempalte
    name = Path(filename).stem
    chords[name] = Image.open(filename)
```

```python
import random
prompt_chords = random.sample(list(chords), 3)
```

```python
from IPython.display import display
from ChordDetect import record, get_chords
import librosa

for name in prompt_chords:
    display(chords[name])
clip, sr = record()
#chords, sr = librosa.load('clips/C_to_G.wav', sr=None)
get_chords(clip, sr)
```
