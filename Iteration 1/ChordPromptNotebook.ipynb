This notebook functions as a final release goal for the First Release and applies all of the chord detection principles researched. First the user is prompted with three images of chords represented as chord diagrams. Once prompted, recording is initiated, the user strums the chords and analysis is performed with functions defined in `ChordDetect.py`

```python
# Source - https://stackoverflow.com/a/26392492
# Posted by user1269942
# Retrieved 2026-10-21, License - CC BY-SA 3.0

from PIL import Image
from pathlib import Path
import glob
# chord dict, so name can be tracked and matched with template
chords = {}
for filename in glob.glob('chords/*.jpg'):
    # need pathlib to get the filename, for analysis use -> matching against tempalte
    name = Path(filename).stem
    chords[name] = Image.open(filename)
```

```python
import random
prompt_chords = random.sample(list(chords), 3)
# prompt_chords = ["A", "G", "E"]
```

```python
from IPython.display import display
from ChordDetect import record, get_chords, check_chords
import librosa

for name in prompt_chords:
    display(chords[name])
clip, sr = record()
# clip, sr = librosa.load('clips/age-humbucker_prompt_test.wav', sr=None)
detected = get_chords(clip, sr)
```

```python
results = check_chords(detected, prompt_chords)
print(results)
```
